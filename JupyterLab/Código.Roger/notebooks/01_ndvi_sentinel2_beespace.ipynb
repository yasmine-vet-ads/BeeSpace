# BeeSpace 01 - NDVI Sentinel-2

Demonstração de cálculo e interpretação de NDVI como indicador de vigor vegetal e disponibilidade potencial de pasto apícola.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

OUTPUT_DIR = Path("outputs")
OUTPUT_DIR.mkdir(exist_ok=True)
```

```python
altura, largura = 120, 120
red = np.random.normal(loc=0.22, scale=0.05, size=(altura, largura))
nir = np.random.normal(loc=0.48, scale=0.10, size=(altura, largura))

nir[30:80, 30:90] += 0.25
red[30:80, 30:90] -= 0.05

red = np.clip(red, 0.01, 1)
nir = np.clip(nir, 0.01, 1)

ndvi = (nir - red) / (nir + red)
ndvi = np.clip(ndvi, -1, 1)

print(f"NDVI mínimo: {ndvi.min():.3f}")
print(f"NDVI médio: {ndvi.mean():.3f}")
print(f"NDVI máximo: {ndvi.max():.3f}")
```

```python
plt.figure(figsize=(7, 6))
plt.imshow(ndvi, vmin=-1, vmax=1)
plt.title("NDVI simulado a partir de bandas Sentinel-2")
plt.colorbar(label="NDVI")
plt.axis("off")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "ndvi_simulado_beespace.png", dpi=150)
plt.show()
```

```python
stats = {
    "ndvi_medio": float(np.mean(ndvi)),
    "ndvi_mediano": float(np.median(ndvi)),
    "ndvi_minimo": float(np.min(ndvi)),
    "ndvi_maximo": float(np.max(ndvi)),
    "perc_pixels_ndvi_alto": float(np.mean(ndvi > 0.50) * 100),
    "perc_pixels_ndvi_baixo": float(np.mean(ndvi < 0.30) * 100)
}
stats_df = pd.DataFrame([stats])
stats_df.to_csv(OUTPUT_DIR / "estatisticas_ndvi_beespace.csv", index=False)
stats_df
```

Interpretação: NDVI alto indica vigor vegetal. A confirmação de florada depende do cruzamento com peso da colmeia, atividade das abelhas e registros de campo.
